Step 1 - Load Data

```python
import numpy as np
import pandas as pd
	
df = pd.read_csv('spam.csv', encoding='latin-1', usecols=[0, 1], dtype=str) # specifying encoding is necessary because the data does not use UTF-8
df.columns = ['Labels', 'SMS']
	
df.head()
```

Step 2 - Preprocessing

```python
#Drop columns that are not used
df = df.drop(df.iloc[:,2:], axis=1)

# Check the data
df.head()
```

```python
	# Check the number of instances per class
	print(df['Labels'].value_counts())
	print('\n')
	
	# Check for completeness of data
	print(df.info())
	print('\n')
	
	# Check descriptive statistics
	print(df.describe())
```

```python
	# Data for labels
	new_labels = {
	    'spam': 1,
	    'ham': 0
	}
	
	# Encode labels
	df['Labels'] = df['Labels'].map(new_labels)
	
	# Check the data
	df.head()
```

```python
	X = df['SMS'].to_numpy(dtype=str)
	y = df['Labels'].to_numpy()
```

Step 3 - Feature Extraction

```python
	from sklearn.model_selection import train_test_split
	from sklearn.feature_extraction.text import CountVectorizer
	
	# Split training and testing data
	X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
	
	# Initialize CountVectorizer
	bow = CountVectorizer()
	
	# Fit and transform X_train with CountVectorizer
	X_train = bow.fit_transform(X_train)
	
	# Transform X_test
	# Why only transform? The same reason as in experiment 3
	# We do not want the model to know the parameters used by CountVectorizer to fit X_train
	# Thus, the testing data can remain unfamiliar to the model later
	X_test = bow.transform(X_test)
```

Step 4 - Model Training and Evaluation

```python
	from sklearn.naive_bayes import MultinomialNB
	from sklearn.metrics import accuracy_score
	
	# Initialize MultinomialNB
	mnb = MultinomialNB()
	
	# Fit the model
	mnb.fit(X_train, y_train)
	
	# Predict on the training data
	y_pred_train = mnb.predict(X_train)
	
	# Evaluate training accuracy
	acc_train = accuracy_score(y_train, y_pred_train)
	
	# Predict on the training data
	y_pred_test = mnb.predict(X_test)
	
	# Evaluate training accuracy
	acc_test = accuracy_score(y_test, y_pred_test)
	
	# Print evaluation results
	print(f'Training data accuracy: {acc_train}')
	print(f'Test data accuracy: {acc_test}')py
```
