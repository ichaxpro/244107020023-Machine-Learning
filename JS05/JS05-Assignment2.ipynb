#### Question 1

Construct a Multinomial Naive Bayes classification model with the following stipulations,
1. Utilizing the dataset spam.csv
2. Feature CountVectorizer by enabling stop_words
3. Evaluate the results

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

df = pd.read_csv('spam.csv', encoding='latin-1', usecols=[0, 1], dtype=str)
df.columns = ['label', 'message']
df['label'] = df['label'].map({'ham': 0, 'spam': 1})
df = df.dropna(subset=['label', 'message'])

X_train, X_test, y_train, y_test = train_test_split(
    df['message'],
    df['label'],
    test_size=0.2,
    random_state=42,
    stratify=df['label']
)
```

```python
# Question 1: Multinomial Naive Bayes with CountVectorizer
count_vectorizer = CountVectorizer(stop_words='english')
X_train_count = count_vectorizer.fit_transform(X_train)
X_test_count = count_vectorizer.transform(X_test)

count_model = MultinomialNB()
count_model.fit(X_train_count, y_train)
count_prediction = count_model.predict(X_test_count)
count_accuracy = accuracy_score(y_test, count_prediction)

print(f'CountVectorizer accuracy: {count_accuracy:.4f}')
print('\nClassification report:')
print(classification_report(y_test, count_prediction, target_names=['ham', 'spam']))
print('Confusion matrix:')
print(confusion_matrix(y_test, count_prediction))
```

#### Question 2

Construct a Multinomial Naive Bayes classification model with the following stipulations,
1. Utilizing the dataset spam.csv
2. Feature TF-IDF by enabling stop_words
3. Evaluate the results and compare them with the outcomes from Assignment no. 2.
4. Provide a conclusion as to which feature is optimal for the given dataset spam.csv

```python
# Question 2: Multinomial Naive Bayes with TF-IDF
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

tfidf_model = MultinomialNB()
tfidf_model.fit(X_train_tfidf, y_train)
tfidf_prediction = tfidf_model.predict(X_test_tfidf)
tfidf_accuracy = accuracy_score(y_test, tfidf_prediction)

print(f'TF-IDF accuracy: {tfidf_accuracy:.4f}')
print('\nClassification report:')
print(classification_report(y_test, tfidf_prediction, target_names=['ham', 'spam']))
print('Confusion matrix:')
print(confusion_matrix(y_test, tfidf_prediction))

comparison = pd.DataFrame({
    'feature': ['CountVectorizer', 'TF-IDF'],
    'accuracy': [count_accuracy, tfidf_accuracy]
})
display(comparison)

best_feature = comparison.loc[comparison['accuracy'].idxmax(), 'feature']
print(f'Conclusion: {best_feature} is optimal for this dataset based on the higher test accuracy.')
```
